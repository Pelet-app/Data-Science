```python
import pandas as pd

df_raw = pd.read_csv('Dataset_pekerjaan.csv')

print("--- 5 Data Teratas Dataset Mentah ---")
display(df_raw.head())
```

## Assessing Data
Kita melakukan evaluasi terhadap struktur kolom yang tersedia dan mendeteksi keberadaan data yang kosong (missing values) sebelum melakukan transformasi.

```python
print("Kolom pada dataset mentah:", df_raw.columns.tolist())
# Memeriksa informasi umum dataset
print("--- Informasi Dataset ---")
df_raw.info()

# Memeriksa jumlah data yang kosong (null) di setiap kolom
print("\n--- Jumlah Missing Values ---")
print(df_raw.isna().sum())

# Memeriksa jumlah data duplikat
print("\n--- Jumlah Data Duplikat ---")
print(f"Data duplikat: {df_raw.duplicated().sum()}")

# Mengecek jumlah data kosong pada 4 kolom target utama
target_columns = ['Kode Unit', 'Judul Unit', 'Elemen Kompetensi', 'Jabatan']
print("\nJumlah missing values pada kolom utama:")
print(df_raw[target_columns].isna().sum())
```

## Cleaning Data dan Penyelarasan Dataset
Pada tahap ini, kita melakukan proses pembersihan data secara manual:
1. Memilih hanya kolom-kolom spesifik yang dibutuhkan (`Kode Unit`, `Judul Unit`, `Elemen Kompetensi`, `Jabatan`).
2. Menghapus baris yang mengandung nilai kosong (*missing values*) pada kolom tersebut.
3. Menghapus baris data yang duplikat untuk memastikan integritas dan keunikan referensi unit kompetensi jabatan.
4. Menyusun urutan kolom agar persis sama dengan format referensi SKKNI.

```python
# Menyeleksi kolom yang diperlukan dan menghapus baris kosong (NaN)
df_cleaned = df_raw[['Kode Unit', 'Judul Unit', 'Elemen Kompetensi', 'Jabatan']].dropna()

# Menghapus baris data yang duplikat keseluruhan
df_cleaned = df_cleaned.drop_duplicates().reset_index(drop=True)

# Memastikan urutan kolom sesuai dengan file referensi bersih
df_cleaned = df_cleaned[['Kode Unit', 'Judul Unit', 'Elemen Kompetensi', 'Jabatan']]

print("Ukuran dataset setelah proses pembersihan (Cleaning):", df_cleaned.shape)
```

## Export Hasil Akhir (Data Terkompresi Siap Pakai)
Menyimpan data hasil pembersihan final ke dalam file CSV baru yang siap digunakan untuk tahapan analisis lanjutan maupun pemodelan.

```python
# Menyimpan data hasil cleaning ke dalam file CSV baru
df_cleaned.to_csv('skkni_reference_clean.csv', index=False)
print("File 'skkni_reference_clean.csv' berhasil diexport dan disimpan!")
```
