## Tahap 1 - Gathering Data

Pada tahap ini dilakukan proses pengumpulan data dengan membaca dataset lowongan pekerjaan dari file `job-vacancy-dataset.csv`.

Dataset kemudian dimuat menggunakan library `pandas` untuk mempermudah proses analisis lebih lanjut.

Output dari tahap ini adalah:

- Mengetahui jumlah kolom (fitur)

Tahap ini merupakan langkah awal dalam data pipeline sebelum dilakukan eksplorasi dan pembersihan data.

```python
import pandas as pd
import re
# GATHERING DATA print
print("TAHAP 1 - GATHERING DATA")
print("Membaca dataset job vacancy...")

df = pd.read_csv( "job-vacancy-dataset.csv", low_memory=False )

print("Dataset berhasil dibaca.")
print(f"Jumlah data : {len(df)}")
print(f"Jumlah kolom : {len(df.columns)}")
```

## Tahap 3 - Cleaning Data

Tahap ini merupakan proses pembersihan data untuk meningkatkan kualitas dataset agar siap digunakan dalam analisis.

### 1. Text Cleaning
Dilakukan pembersihan pada kolom teks (`title`, `requirement`, `description`) dengan langkah:
- Mengubah teks menjadi lowercase
- Menghapus karakter non-ASCII
- Menghapus URL
- Menghapus simbol dan tanda baca
- Merapikan spasi berlebih

### 2. Filtering Data
Dataset difilter berdasarkan kata kunci pekerjaan tertentu seperti:
- Cyber Security
- Software Engineer
- Data Analyst
- Accounting
- Electrical Engineer

Tujuan filtering ini adalah untuk fokus pada domain pekerjaan yang relevan dengan project capstone.

### 3. Cleaning Data Numerik
Kolom numerik seperti:
- salary
- company size
- work experience

dibersihkan dengan cara:
- Mengubah ke format numerik
- Mengisi nilai kosong dengan 0

### 4. Handling Missing Value
Seluruh missing value pada kolom non-numerik diganti dengan nilai `"None"` agar tidak ada data kosong dalam dataset.

### 5. Cleaning Time Column
Kolom waktu dibersihkan dari karakter tidak valid dan nilai error seperti `###`.

```python
# CLEANING DATA
print("TAHAP 3 - CLEANING DATA")

# 1. FUNGSI CLEAN TEXT
print("Membuat fungsi clean_text...")
def clean_text(text):
    if pd.isna(text) or not isinstance(text, str) or str(text).strip() == "":
        return "None"

    # Lowercase
    text = text.lower()

    # Hapus karakter non-ascii
    text = text.encode('ascii', 'ignore').decode('ascii')

    # Hapus URL
    text = re.sub(r'https?://\s*\S+|www\.\S+', '', text)

    # Hapus simbol
    text = re.sub(r'[^\w\s]', ' ', text)

    # Rapihkan spasi
    text = re.sub(r'\s+', ' ', text).strip()
    return text if text != "" else "None"
print("Fungsi clean_text berhasil dibuat.\n")

# 2. FILTER PEKERJAAN TARGET
print("Filtering pekerjaan target...")
keywords_pattern = (
    r'cyber\s*security|information\s*security|network\s*security|penetration\s*tester|keamanan\s*informasi|'
    r'staff\s*akuntansi|accounting\s*staff|accounting\s*officer|teknisi\s*akuntansi|junior\s*accountant|'
    r'\bprogrammer\b|software\s*engineer|web\s*developer|backend\s*developer|frontend\s*developer|fullstack\s*developer|'
    r'data\s*analyst|data\s*analytics|junior\s*data\s*analyst|'
    r'insinyur\s*elektro|electrical\s*engineer|teknisi\s*elektro|teknisi\s*listrik'
)

df['title'] = df['title'].astype(str)
df = df[
    df['title']
    .str.lower()
    .str.contains(keywords_pattern, na=False, regex=True)
].copy()
print("Filtering pekerjaan berhasil.")
print(f"Jumlah data setelah filtering : {len(df)}\n")

# 3. CLEANING KOLOM TEKS
print("Membersihkan kolom teks...")
text_columns = ['title', 'requirement', 'description']
for col in text_columns:
    if col in df.columns:
        df[col] = df[col].apply(clean_text)
print("Kolom teks berhasil dibersihkan.\n")

# 4. CLEANING DATA NUMERIK
print("Membersihkan data numerik...")
numeric_columns = [
    'min_company_size',
    'max_company_size',
    'min_work_experience',
    'max_work_experience',
    'min_salary',
    'max_salary'
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = (
            pd.to_numeric(df[col], errors='coerce')
            .fillna(0)
            .astype(int)
        )
print("Data numerik berhasil dibersihkan.\n")

# 5. CLEANING MISSING VALUES
print("Membersihkan missing values...")
for col in df.columns:
    if col not in numeric_columns:
        df[col] = df[col].fillna('None')
        df[col] = df[col].astype(str)
        df[col] = df[col].replace(
            r'^\s*$',
            'None',
            regex=True
        )
        df[col] = df[col].replace(
            ['nan', 'NaN'],
            'None'
        )

print("Missing values berhasil dibersihkan.\n")

# 6. CLEANING KOLOM WAKTU
print("Membersihkan kolom waktu...")
if 'time' in df.columns:
    df['time'] = df['time'].astype(str)
    df['time'] = df['time'].str.replace(
        r'#+',
        'None',
        regex=True
    )

print("Kolom waktu berhasil dibersihkan.\n")
```

Hasil akhir cleaning disimpan ke file baru `cleaned_job_5.csv` untuk digunakan pada tahap analisis atau modeling selanjutnya.

```python
# MENYIMPAN HASIL AKHIR
print("Menyimpan hasil akhir...")

output_file = "cleaned_job_5.csv"

df.to_csv(output_file, index=False)

print("DATASET BERHASIL DIBERSIHKAN")
print(f"Output tersimpan di: {output_file}")
```
