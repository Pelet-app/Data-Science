# Data Wrangling
### Persiapan Data
Dokumen ini berisi tahapan Data Wrangling yang dilakukan untuk mempersiapkan dataset sebelum proses analisis dan pemodelan. Dataset yang digunakan terdiri dari data lowongan pekerjaan, peran pekerjaan, keterampilan, dan resume pelamar kerja.

Proses Data Wrangling meliputi pengumpulan data (gathering), pemeriksaan kualitas data (assessing), serta pembersihan data (cleaning) untuk memastikan data yang digunakan lebih konsisten dan siap diolah pada tahap berikutnya.

```python
# Import library
import pandas as pd
import numpy as np
import re
```

### Gathering Data
Pada tahap ini dilakukan pemuatan seluruh dataset yang akan digunakan dalam proyek. Dataset dibaca menggunakan library Pandas dan disimpan ke dalam DataFrame untuk memudahkan proses pengolahan data selanjutnya.

```python
jobs = pd.read_csv("all_job_post.csv")
roles = pd.read_csv("job_roles.csv")
skills = pd.read_csv("skills_list.csv")
resume = pd.read_csv("training_data.csv")
```

```python
# Menampilkan Ukuran Dataset
# Melihat jumlah baris dan kolom setiap dataset
print("Jobs Dataset :", jobs.shape)
print("Roles Dataset :", roles.shape)
print("Skills Dataset :", skills.shape)
print("Resume Dataset :", resume.shape)
```

# Assesing Data
Tahap assessing dilakukan untuk memahami struktur dataset serta mengidentifikasi potensi permasalahan data. Pemeriksaan meliputi jumlah data, tipe data, missing values, dan data duplikat yang dapat memengaruhi kualitas analisis.

```python
# Dataset jobs
jobs.info() # Melihat informasi dataset
jobs.isnull().sum() # Mengecek missing values
jobs.duplicated().sum() # Mengecek data duplikat
jobs.head() # Menampilkan beberapa data awal
```

```python
# Dataset roles
roles.info()
roles.isnull().sum()
roles.duplicated().sum()
roles.head()
```

```python
# Dataset Skills
skills.info()
skills.isnull().sum()
skills.duplicated().sum()
skills.head()
```

```python
# Dataset resume
resume.info()
resume.isnull().sum()
resume.duplicated().sum()
resume.head()
```

# Cleaning Data
Berdasarkan hasil pemeriksaan sebelumnya, dilakukan proses pembersihan data untuk meningkatkan kualitas dataset. Proses ini mencakup penghapusan data duplikat, penanganan missing values, serta normalisasi teks agar format data menjadi lebih konsisten.

```python
# Fungsi pembersihan teks
def clean_text(text):
    if pd.isna(text):
        return ""
    text = text.lower()
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"[^a-zA-Z\s]", "", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()
```

```python
# Membersihkan Dataset Jobs

# Menghapus Duplikasi dan Missing Value
jobs = jobs.drop_duplicates()
jobs = jobs.dropna()

# Membersihkan Kolom Deskripsi dan Judul Pekerjaan
jobs["clean_desc"] = jobs["job_description"].apply(clean_text)
jobs["clean_title"] = jobs["job_title"].apply(clean_text)

jobs.head()
```

```python
# Membersihkan Dataset Resume

# Menghapus Duplikasi dan Missing Value
resume = resume.drop_duplicates()
resume = resume.dropna()

# Membersihkan Isi Resume
resume["clean_resume"] = resume["Resume Text"].apply(clean_text)

resume.head()
```

```python
# Membersihkan Dataset Skills

# Menghapus Duplikasi dan Missing Value
skills = skills.drop_duplicates()
skills = skills.dropna()

# Menyeragamkan Format Skill
skills["clean_skill"] = skills["Skill Name"].str.lower().str.strip()

skills.head()
```

```python
# Membersihkan Dataset Roles

# Menghapus Duplikasi dan Missing Value
roles = roles.drop_duplicates()
roles = roles.dropna()

# Menyeragamkan Format Nama Role
roles["clean_role"] = roles["Job Title"].str.lower().str.strip()

roles.head()
```

# Verifikasi Hasil Cleaning

```python
# Dataset Jobs
print("Missing Value Jobs:")
print(jobs.isnull().sum())

print("\nDuplicate Jobs:")
print(jobs.duplicated().sum())
```

```python
# Dataset Roles
print("Missing Value Roles:")
print(roles.isnull().sum())

print("\nDuplicate Roles:")
print(roles.duplicated().sum())
```

```python
# Dataset Skills
print("Missing Value Skills:")
print(skills.isnull().sum())

print("\nDuplicate Skills:")
print(skills.duplicated().sum())
```

```python
# Dataset Resume
print("Missing Value Resume:")
print(resume.isnull().sum())

print("\nDuplicate Resume:")
print(resume.duplicated().sum())
```

# Menyimpan Dataset Hasil Cleaning

```python
# Menyimpan dataset yang telah dibersihkan
jobs.to_csv("cleaned_all_jobs_post.csv", index=False)
resume.to_csv("cleaned_training_data.csv", index=False)
skills.to_csv("cleaned_skills_list.csv", index=False)
roles.to_csv("cleaned_job_roles.csv", index=False)
```
